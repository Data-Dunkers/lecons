```python
import pandas as pd
sheet_url = "https://docs.google.com/spreadsheets/d/11kzLuW9jlhvjzLhNeTntfaJ4AAdtHxBfKhRYuM8Ynsc/edit?usp=sharing"
csv_url = sheet_url.split("edit")[0] + "export?format=csv"
print(csv_url)
df = pd.read_csv(csv_url)

print(f'Il y a {df.shape[0]} rangées et {df.shape[1]} colonnes de données\n')
for i, column in enumerate(df.columns):
    print(i, column)
```

```python
x_column = 9
y_column = 14

# do not edit below this line
import plotly.express as px
x = df.columns[x_column]
y = df.columns[y_column]
px.scatter(df, x=x, y=y, title=y+' vs '+x, trendline='ols')
```

## Carte de correlation

Nous pouvons également créer une carte de corrélation (utilisant  [couleurs](https://matplotlib.org/stable/gallery/color/colormap_reference.html)) pour voir comment les variables numériques sont liées.

```python
numeric_df = df.select_dtypes(include=['number'])
numeric_df.drop('Timestamp', axis=1, errors='ignore').drop('Quel numéro de feuille as-tu?', axis=1).corr().style.background_gradient(cmap='cividis').format(precision=3)
```
